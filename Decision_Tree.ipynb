```python
import pandas as pd
import numpy as np
import time

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder

from sklearn.tree import DecisionTreeClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)
```

```python
dataset = pd.read_csv("/content/dataset2_training.csv")

print(dataset.shape)
dataset.head()
```

```python
X = dataset.drop("Disease", axis=1)
y = dataset["Disease"]

print("Features:", X.shape)
print("Labels:", y.shape)
```

```python
encoder = LabelEncoder()

y = encoder.fit_transform(y)

print("Total Diseases:", len(np.unique(y)))
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print(X_train.shape)
print(X_test.shape)
```

```python
start = time.time()

dt = DecisionTreeClassifier(
    random_state=42
)

dt.fit(X_train, y_train)

end = time.time()
```

```python
y_pred = dt.predict(X_test)
```

```python
accuracy = accuracy_score(y_test, y_pred)

precision = precision_score(
    y_test,
    y_pred,
    average="weighted",
    zero_division=0
)

recall = recall_score(
    y_test,
    y_pred,
    average="weighted",
    zero_division=0
)

f1 = f1_score(
    y_test,
    y_pred,
    average="weighted",
    zero_division=0
)

print("Decision Tree Accuracy :", accuracy)
print("Precision :", precision)
print("Recall :", recall)
print("F1 Score :", f1)
print("Training Time :", round(end-start,2),"seconds")
```

```python
print(classification_report(
    y_test,
    y_pred,
    zero_division=0
))
```

```python
cm = confusion_matrix(y_test, y_pred)

print(cm.shape)
```

```python
import joblib

joblib.dump(dt, "DecisionTree_Model.pkl")

print("Model Saved Successfully")
```

```python
results = pd.DataFrame({
    "Model":["Decision Tree"],
    "Accuracy":[accuracy],
    "Precision":[precision],
    "Recall":[recall],
    "F1 Score":[f1],
    "Training Time":[round(end-start,2)]
})

results
```
