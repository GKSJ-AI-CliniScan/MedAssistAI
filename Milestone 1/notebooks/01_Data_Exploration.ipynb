```python
# Empty list to store transformed rows
transformed_rows = []

# Loop through every row
for _, row in diabetes_df.iterrows():

    # Create empty row
    new_row = pd.Series(index=disease_df.columns, dtype="object")
    new_row.iloc[:-1] = 0

    # Disease name
    new_row["prognosis"] = row["Disease"].strip()

    # Symptoms
    for symptom in row[1:]:

        if pd.notna(symptom) and symptom != "None":

            symptom = symptom.strip()   # ⭐ Remove leading/trailing spaces

            if symptom in disease_df.columns:
                new_row[symptom] = 1

    transformed_rows.append(new_row)

# Create DataFrame
diabetes_transformed = pd.DataFrame(transformed_rows)

print(diabetes_transformed.shape)
```

```python
# Load processed datasets

disease_df = pd.read_csv("../processed_data/disease_clean.csv")
diabetes_df = pd.read_csv("../processed_data/diabetes_transformed.csv")
patient_df = pd.read_csv("../processed_data/patient_clean.csv")
```

```python
print("Disease:", disease_df.shape)
print("Diabetes:", diabetes_df.shape)
print("Patient:", patient_df.shape)
```

```python
def clean_disease_name(name):
    if pd.isna(name):
        return name

    name = str(name).strip().lower()

    replacements = {
        "diabetes ": "diabetes",
        "hypertension ": "hypertension",
        "hepatitis a": "hepatitis a",
        "chicken pox": "chickenpox",
        "urinary tract infection": "urinary tract infection",
        "typhoid": "typhoid fever",
        "heart attack": "myocardial infarction (heart attack)",
        "bronchial asthma": "asthma",
        "common cold": "common cold",
        "osteoarthristis": "osteoarthritis"
    }

    return replacements.get(name, name)
```

```python
disease_df["prognosis"] = disease_df["prognosis"].apply(clean_disease_name)

diabetes_df["prognosis"] = diabetes_df["prognosis"].apply(clean_disease_name)

patient_df["Disease"] = patient_df["Disease"].apply(clean_disease_name)

print("Done!")
```

```python
common = set(disease_df["prognosis"]) & set(diabetes_df["prognosis"]) & set(patient_df["Disease"])

print("Common diseases:", len(common))
print(sorted(common))
```

```python
# Merge both datasets

final_dataset = pd.concat(
    [disease_df, diabetes_df],
    ignore_index=True
)

print("Shape before removing duplicates:", final_dataset.shape)
```

```python
duplicates = final_dataset.duplicated().sum()

print("Duplicate rows:", duplicates)
```

```python
# Remove duplicate rows

final_dataset = final_dataset.drop_duplicates()

print("Final Shape:", final_dataset.shape)
```

```python
final_dataset.to_csv(
    "../processed_data/final_dataset.csv",
    index=False
)

print("✅ Final dataset saved successfully!")
```
