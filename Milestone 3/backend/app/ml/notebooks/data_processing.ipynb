```python
# 1. Count the number of unique disease names in the text column
num_diseases = df_clean['diseases'].nunique()

# 2. Count the symptoms (Total columns minus 1 for the 'diseases' column)
num_symptoms = df_clean.shape[1] - 1

print(f"Total Unique Diseases: {num_diseases}")
print(f"Total Symptoms: {num_symptoms}")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(15,5))
class_counts = df_clean['diseases'].value_counts()
class_counts.sort_values(ascending=False).plot(kind='bar')

plt.title("Number of Samples per Disease")
plt.xlabel("Disease")
plt.ylabel("Number of Samples")
plt.show()
```

```python
bins = [0, 3, 10, 30, 100, 300, 600, 900, float('inf')]
labels = [
    "1-3",
    "4-10",
    "11-30",
    "31-100",
    "101-300",
    "301-600",
    "601-900",
    "901+"
]

groups = pd.cut(class_counts, bins=bins, labels=labels)

print(groups.value_counts().sort_index())
```

```python
for cap in [800, 600, 500, 400, 300]:
    new_rows = class_counts.clip(upper=cap).sum()
    removed = len(df_clean) - new_rows

    print(
        f"Cap {cap}: "
        f"Rows after = {new_rows}, "
        f"Removed = {removed}"
    )
```
