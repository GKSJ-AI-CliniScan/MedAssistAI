```python
import pandas as pd
import numpy as np
import time
import warnings

warnings.filterwarnings("ignore")

print("Libraries imported successfully.")
```

```python
dataset = pd.read_csv("/content/dataset2_training.csv")

print("Dataset loaded successfully.")
print("Dataset Shape:", dataset.shape)
print("Number of Features:", dataset.shape[1] - 1)
print("Number of Diseases:", dataset["Disease"].nunique())
```

```python
print("Columns:", dataset.columns.tolist()[:10])

print("\nTarget column exists:",
      "Disease" in dataset.columns)

print("\nMissing values in Disease:",
      dataset["Disease"].isna().sum())

print("\nUnique diseases:",
      dataset["Disease"].nunique())
```

```python
X = dataset.drop(columns=["Disease"])
y = dataset["Disease"]

print("X Shape:", X.shape)
print("y Shape:", y.shape)
```

```python
print("Non-numeric columns:")

non_numeric_columns = X.select_dtypes(
    exclude=["number"]
).columns.tolist()

print(non_numeric_columns)
```

```python
missing_values = X.isna().sum().sum()

print("Total missing values:", missing_values)
```

```python
from sklearn.preprocessing import LabelEncoder

label_encoder_gb = LabelEncoder()

y_encoded = label_encoder_gb.fit_transform(y)

print("Total Classes:",
      len(label_encoder_gb.classes_))

print("Minimum Label:",
      y_encoded.min())

print("Maximum Label:",
      y_encoded.max())
```

```python
from sklearn.model_selection import train_test_split

X_train_gb, X_test_gb, y_train_gb, y_test_gb = train_test_split(
    X,
    y_encoded,
    test_size=0.20,
    random_state=42,
    stratify=y_encoded
)

print("X_train:", X_train_gb.shape)
print("X_test :", X_test_gb.shape)

print("y_train:", y_train_gb.shape)
print("y_test :", y_test_gb.shape)

print(
    "Train classes:",
    len(np.unique(y_train_gb))
)

print(
    "Test classes:",
    len(np.unique(y_test_gb))
)
```

```python
missing_train_classes = (
    set(range(658))
    - set(np.unique(y_train_gb))
)

missing_test_classes = (
    set(range(658))
    - set(np.unique(y_test_gb))
)

print(
    "Missing classes in training:",
    missing_train_classes
)

print(
    "Number of classes missing in testing:",
    len(missing_test_classes)
)
```

```python
from sklearn.ensemble import GradientBoostingClassifier

print("Gradient Boosting imported successfully.")
```

```python
gb_model = GradientBoostingClassifier(
    n_estimators=20,
    learning_rate=0.1,
    max_depth=3,
    min_samples_leaf=2,
    random_state=42
)

print(gb_model)
```

```python
start_gb_training = time.time()

gb_model.fit(
    X_train_gb,
    y_train_gb
)

end_gb_training = time.time()

gb_training_time = (
    end_gb_training
    - start_gb_training
)

print(
    "Gradient Boosting Training Time:",
    round(gb_training_time, 2),
    "seconds"
)
```

```python
start_gb_prediction = time.time()

gb_pred = gb_model.predict(
    X_test_gb
)

end_gb_prediction = time.time()

gb_prediction_time = (
    end_gb_prediction
    - start_gb_prediction
)

print(
    "Gradient Boosting Prediction Time:",
    round(gb_prediction_time, 2),
    "seconds"
)

print(
    "Unique Predictions:",
    len(np.unique(gb_pred))
)
```

```python
from sklearn.metrics import accuracy_score

gb_accuracy = accuracy_score(
    y_test_gb,
    gb_pred
)

print(
    "Gradient Boosting Accuracy:",
    round(gb_accuracy, 4)
)
```

```python
from sklearn.metrics import (
    precision_score,
    recall_score,
    f1_score
)

gb_precision = precision_score(
    y_test_gb,
    gb_pred,
    average="weighted",
    zero_division=0
)

gb_recall = recall_score(
    y_test_gb,
    gb_pred,
    average="weighted",
    zero_division=0
)

gb_f1 = f1_score(
    y_test_gb,
    gb_pred,
    average="weighted",
    zero_division=0
)

print("Gradient Boosting Results")
print("-------------------------")
print("Accuracy :", round(gb_accuracy, 4))
print("Precision:", round(gb_precision, 4))
print("Recall   :", round(gb_recall, 4))
print("F1 Score :", round(gb_f1, 4))
print("Training Time:", round(gb_training_time, 2), "seconds")
print("Unique Predictions:", len(np.unique(gb_pred)))
```

```python
from sklearn.metrics import classification_report

gb_report = classification_report(
    y_test_gb,
    gb_pred,
    zero_division=0
)

print(gb_report)
```
