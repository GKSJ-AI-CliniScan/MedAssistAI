```python
## Duplicate Record Investigation

Before removing duplicate records, the datasets were inspected to determine whether the duplicates were introduced intentionally (e.g., during data augmentation) or were accidental duplicates.

This investigation helps ensure that useful training data is not removed unnecessarily.
```

```python
## Duplicate Handling

### Dataset 1
- No duplicate records were found.

### Dataset 2
- Duplicate records were identified.
- The dataset is an augmented dataset, so duplicate rows were investigated before deciding whether to remove them.

### Dataset 3
- Exact duplicate records were removed because they did not contribute additional information.
```

```python
dataset2_clean = dataset2_clean.drop_duplicates().reset_index(drop=True)
```

```python
## Duplicate Removal

### Dataset 2
- 57,298 duplicate records were identified.
- Exact duplicate rows were removed.
- Remaining records: 189,647.

### Dataset 3
- 28,988 duplicate records were identified.
- Exact duplicate rows were removed.
- Remaining records: 163,727.

Removing duplicate records helps reduce redundancy and prevents model bias during training.
```

```python
description_clean["Description"] = (
    description_clean["Description"]
    .str.strip()
    .str.replace(r"\s+", " ", regex=True)
)
```

```python
diet_clean["Diet"] = (
    diet_clean["Diet"]
    .str.strip()
    .str.replace(r"\s+", " ", regex=True)
)
```

```python
workout_clean["Workouts"] = (
    workout_clean["Workouts"]
    .str.strip()
    .str.replace(r"\s+", " ", regex=True)
)
```

```python
dataset3_clean["symptom_text"] = (
    dataset3_clean["symptom_text"]
    .str.strip()
    .str.replace(r"\s+", " ", regex=True)
)
```
