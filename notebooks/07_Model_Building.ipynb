```python
# Machine Learning Model Building

## Objective

The objective of this notebook is to train, evaluate, and compare multiple machine learning models for disease prediction based on patient symptoms.

### Dataset Used

- dataset2_clean.csv

### Algorithms

- Decision Tree
- Random Forest

### Evaluation Metrics

- Accuracy
- Precision
- Recall
- F1-Score
- Confusion Matrix
```

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split

from sklearn.preprocessing import LabelEncoder

from sklearn.tree import DecisionTreeClassifier

from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import accuracy_score

from sklearn.metrics import classification_report

from sklearn.metrics import confusion_matrix
```

```python
dataset2 = pd.read_csv("../cleaned_datasets/dataset2_clean.csv")
```

```python
dataset2.head()
```

```python
X = dataset2.drop("Disease", axis=1)

y = dataset2["Disease"]
```

```python
print(X.shape)

print(y.shape)
```

```python
from sklearn.preprocessing import LabelEncoder

label_encoder = LabelEncoder()

y_encoded = label_encoder.fit_transform(y)
```

```python
print(y.head())
```

```python
print(y_encoded[:10])
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y_encoded,
    test_size=0.2,
    random_state=42
)
```

```python
print("X_train:", X_train.shape)
print("X_test :", X_test.shape)

print("y_train:", y_train.shape)
print("y_test :", y_test.shape)
```

```python
from sklearn.tree import DecisionTreeClassifier

decision_tree = DecisionTreeClassifier(random_state=42)

decision_tree.fit(X_train, y_train)
```

```python
y_pred_dt = decision_tree.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score

dt_accuracy = accuracy_score(y_test, y_pred_dt)

print("Decision Tree Accuracy:", dt_accuracy)
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred_dt))
```

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test, y_pred_dt)

print(cm)
```

```python
## Decision Tree Model - Observations

- The Decision Tree model was trained using the cleaned disease dataset.
- The dataset was split into 80% training and 20% testing.
- The model achieved an accuracy of approximately **75.5%**.
- The classification report shows strong performance for many diseases, while some rare diseases have lower precision and recall due to fewer training samples.
- The confusion matrix was generated to evaluate prediction performance across all disease classes.
```
