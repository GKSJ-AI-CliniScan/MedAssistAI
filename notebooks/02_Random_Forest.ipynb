```python
import pandas as pd
import joblib

from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
```

```python
# Load Dataset
df = pd.read_csv("../datasets/clean_190k_dataset.csv")

print("Dataset Loaded Successfully!")
print(df.shape)
```

```python
# Reduce dataset size to avoid MemoryError
df = df.sample(frac=0.25, random_state=42)

print("Sampled Dataset Shape:")
print(df.shape)
```

```python
X = df.drop("diseases", axis=1)

y = df["diseases"]

print("Features:", X.shape)
print("Target:", y.shape)
```

```python
label_encoder = LabelEncoder()

y_encoded = label_encoder.fit_transform(y)

print("Number of Disease Classes:", len(label_encoder.classes_))
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y_encoded,
    test_size=0.2,
    random_state=42
)

print("Training Set:", X_train.shape)
print("Testing Set:", X_test.shape)
```

```python
rf_model = RandomForestClassifier(
    n_estimators=30,
    max_depth=15,
    max_features="sqrt",
    random_state=42,
    n_jobs=1
)
```

```python
print("Training Started...")

rf_model.fit(X_train, y_train)

print("Model Trained Successfully!")
```

```python
joblib.dump(rf_model, "../models/random_forest_model.pkl")

joblib.dump(label_encoder, "../models/label_encoder.pkl")

print("Model Saved Successfully!")
```

## Hardware Limitation

The Random Forest model could not be trained successfully on the complete
dataset (189,647 records and 773 disease classes) using the available
hardware (8 GB RAM).

During training, a MemoryError occurred because Random Forest constructs
multiple decision trees simultaneously, requiring more memory than was
available.

The implementation is correct and can be trained successfully on a machine
with higher RAM (16 GB or more) or on cloud computing platforms.
