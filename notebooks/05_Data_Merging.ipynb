```python
# Data Merging

## Objective

The objective of this notebook is to merge all cleaned knowledge-base datasets into a single disease information dataset.

### Datasets Used

- disease_clean.csv
- description_clean.csv
- medications_clean.csv
- precautions_clean.csv
- diet_clean.csv
- workout_clean.csv

### Output

knowledge_base.csv
```

```python
import pandas as pd
import os
```

```python
print(os.getcwd())
```

```python
# Load cleaned knowledge base datasets

disease_df = pd.read_csv("../cleaned_datasets/disease_clean.csv")
description_df = pd.read_csv("../cleaned_datasets/description_clean.csv")
medications_df = pd.read_csv("../cleaned_datasets/medications_clean.csv")
precautions_df = pd.read_csv("../cleaned_datasets/precautions_clean.csv")
diet_df = pd.read_csv("../cleaned_datasets/diet_clean.csv")
workout_df = pd.read_csv("../cleaned_datasets/workout_clean.csv")
```

```python
print("Disease Dataset:", disease_df.shape)
print("Description Dataset:", description_df.shape)
print("Medication Dataset:", medications_df.shape)
print("Precaution Dataset:", precautions_df.shape)
print("Diet Dataset:", diet_df.shape)
print("Workout Dataset:", workout_df.shape)
```

```python
print(disease_df.columns.tolist())
print(description_df.columns.tolist())
print(medications_df.columns.tolist())
print(precautions_df.columns.tolist())
print(diet_df.columns.tolist())
print(workout_df.columns.tolist())
```

```python
print("Disease Dataset:", disease_df["Disease"].nunique())
print("Description:", description_df["Disease"].nunique())
print("Medication:", medications_df["Disease"].nunique())
print("Precautions:", precautions_df["Disease"].nunique())
print("Diet:", diet_df["Disease"].nunique())
print("Workout:", workout_df["Disease"].nunique())
```

```python
## Merge Disease Dataset with Description

The disease dataset is merged with the description dataset using the Disease column. This allows every disease record to include its corresponding description.
```

```python
knowledge_base = pd.merge(
    disease_df,
    description_df,
    on="Disease",
    how="left"
)
```

```python
knowledge_base.shape
```

```python
knowledge_base.head()
```

```python
## Merge Medication Information

Medication information is added to the knowledge base using the Disease column.
```

```python
knowledge_base = pd.merge(
    knowledge_base,
    medications_df,
    on="Disease",
    how="left"
)
```

```python
knowledge_base = pd.merge(
    knowledge_base,
    precautions_df,
    on="Disease",
    how="left"
)
```

```python
knowledge_base = pd.merge(
    knowledge_base,
    diet_df,
    on="Disease",
    how="left"
)
```

```python
knowledge_base = pd.merge(
    knowledge_base,
    workout_df,
    on="Disease",
    how="left"
)
```
